```python
df = pd.read_csv('car_fuel_efficiency_2026.csv')
```

```python
df
```

```python
asia = df[df['origin'] == 'Asia']
```

```python
asia_selected = asia[['vehicle_weight', 'model_year']]
```

```python
asia_selected
```

```python
asia_first_7 = asia_selected.head(7)
```

```python
asia_first_7
```

```python
X = asia_first_7.to_numpy()
```

```python
X.shape
```

транспонируем мартрицу

```python
X.T
```

```python
XTX = X.T @ X
```

```python
XTX
```

Находим обратную матрицу(нужна для формулы линейной регрессии) 

```python
XTX_inv = np.linalg.inv(XTX)
```

Создаем массив Y

```python
y = np.array([1100, 1300, 800, 900, 1000, 1100, 1200])
```

```python
w = XTX_inv @ X.T @ y
```
